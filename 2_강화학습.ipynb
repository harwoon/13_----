# **1. 지도학습과 강화학습**

### 1.  지도학습 (Supervised Learning)
지도학습은 정답(라벨)이 포함된 데이터로 모델을 학습시키는 방식입니다. 즉, 입력 X와 그에 대응하는 정답 Y를 함께 제공하여, 모델이 X→Y 매핑하는 함수를 학습하도록 합니다. 학습 과정에서는 모델의 예측값과 실제 정답 사이의 오차(loss)를 계산하고, 이를 최소화하도록 가중치를 업데이트합니다. 대표적인 문제 유형은 분류(Classification)와 회귀(Regression)이며, 이미지 분류, 스팸 메일 판별, 주가 예측 등이 이에 해당합니다. 핵심은 "정답이 있는 데이터로 오차를 줄이며 학습한다"는 점입니다.

### 2. 강화학습 (Reinforcement Learning)
강화학습은 정답을 직접 주지 않고, 행동에 대한 보상(Reward)을 통해 학습하는 방식입니다. 에이전트(agent)가 환경(environment)과 상호작용하며 행동(action)을 선택하고, 그 결과로 받은 보상을 최대화하는 방향으로 정책(policy)을 학습합니다. 즉, "이 행동이 좋았는가?"를 보상으로 평가받으며 점점 더 좋은 전략을 찾아갑니다. 게임 AI, 자율주행, 로봇 제어 등이 대표적인 예입니다. 핵심은 "정답 대신 보상을 통해 최적 행동 전략을 학습한다"는 점입니다.

### 1. 보상(Reward)
- 에이전트가 특정 상태에서 어떤 행동을 했을 때 환경으로부터 받는 즉각적인 피드백 신호
- 행동의 좋고 나쁨을 수치로 표현한 값으로, 에이전트는 이 리워드를 기반으로 자신의 정책을 업데이트하며 장기적인 누적 보상을 최대화하도록 학습
- 매 타임스탭마다 주어질 수도 있고, 에피소드 종료 시 한 번만 주어질 수도 있으며, 양수는 바람직한 행동을, 음수는 바람직하지 않은 행동을 의미하는 경우가 일반적
    - 즉각적 피드백: 특정 상태에서 행동을 했을 때 즉시 주어지는 수치형 피드백
    - 희소성: 리워드는 매 순간 주어지지 않을 수 있음. 많은 환경에서 대부분의 시간 단계에서 보상이 0이고, 특정 목표를 달성했을 때만 보상이 주어질 수 있음 > 희소할수록 학습이 어려워짐
    - 지연성: 어떤 행동의 결과에 대한 보상이 즉시 나타나지 않고, 여러 단계를 거친 뒤 나중에 주어질 수 있음. 이로 인해 어떤 행동이 최종 결과에 기여했는지를 파악하기 어려워지며, 이를 신용 할당 문제(Credit Assignment Problem)라고 함.
    
- 강화학습의 목표가 단순히 현재 리워드를 최대화하는 것이 아니라, 할인율을 고려한 미래 보상까지 포함한 누적 리워드를 최대화하는 것이 중요

### 2. 에이전트(Agent)
- 환경과 상호작용하며 상태를 관찰하고, 그에 따라 행동을 선택하고, 그 결과로 리워드를 받아 학습하는 주체를 의미
- 에이전트의 목표는 단기적인 보상이 아니라 미래 보상을 포함한 누적 보상을 최대화하는 정책을 학습하는 것
- 경험을 통해 가치 함수나 정책을 점진적으로 업데이트하며, 탐험과 활용을 균형있게 수행

1. 가치 기반 에이전트
    - 가치 함수를 학습하여 행동을 선택하는 방식
    - 각 상태 또는 상태-행동 쌍의 가치를 계산하고, 가치가 가장 높은 행동을 선택
2. 정책 기반 에이전트
    - 행동 선택 규칙인 정책을 직접 학습하는 방식
    - 확률적으로 행동을 출력하는 정책을 최적화함
3. 액터-크리틱 에이전트
    - 가치 기반과 정책 기반을 결합한 방식
    - Actor는 정책을 학습하여 행동을 선택
    - Critic은 가치 함수를 학습하여 Actor의 행동을 평가

### 3. 탐험
- 탐험(Exlploration)이란, 에이전트가 현재까지 학습한 최적 행동만 반복하는 것이 아니라, 더 나은 보상을 얻을 가능성이 있는 새로운 행동을 시도해보는 과정을 의미
- 장기적인 누적 보상을 최대화하기 위해 필수적인 전략으로, 이미 잘 알려진 행동을 선택하는 활용과 균형을 이루어야 함

### 4. 활용
- 활용(Exploitation)이란, 에이전트가 지금까지의 학습 결과를 바탕으로 현재 가장 높은 보상을 줄 것으로 예상되는 행동을 선택하는 과정을 의미
- 이미 경험을 통해 가치가 높다고 판단된 행동을 반복하여 안정적으로 보상을 얻는 전략
- 단기적으로 높은 성과를 보장하지만, 새로운 가능성을 탐색하지 않기 때문에 더 나은 전략을 발견하지 못하고 지역 최적해에 머무를 위험이 있음

# **2. 강화학습을 이 게임에 대입**

### 1. 주요 용어
1. Agent: 선택하는 주체. 케릭터
2. Environment: 행동에 반응하는 세계. 미로와 벽
3. State: 의사결정에 필요한 현재 정보. 케릭터의 위치, 열쇠 보유 여부
4. Observation: 실제로 볼 수 있는 정보. 화면의 일부
5. Action: 선택 가능한 조작. 위/아래/좌/우
6. Reward: 한 단계에서 받은 점수. (예: 도착(+10), 충돌(-1))
7. Policy: 상황별 행동 규칙. 갈림길에서 오른쪽 선택
8. Return: 앞으로 받을 보상들의 합. 여러 이동의 총 평가 점수
9. Episode: 시작부터 종료까지 한 판. 출발 > 도착/실패


### 2. 보상
- 즉시 보상: 동전을 주우면 +1
- 지연 보상: 바둑은 여러 수를 둔 뒤 승리했을 때 +1점
- 희소 보상: 미로 출구에 도착하기 전까지 0점
- 보상 설계의 함정: 앞으로 잘 이동할 때마다 +1을 주면 출구 대신 계속 왔다갔다 이동만 할 수 있음

> 순차적 의사결정: 지금 장애물을 피하기 위해 돌아가는 행동은 당장은 느리지만, 나중에 목적지에 안전하게 도착할 수 있음

> ε-greedy: 10%는 무작위로 탐험. 나머지 90%는 현재 가장 좋다고 추정한 행동 선택. 최고는 지금까지의 추정치이지 진짜 최고가 아님

# **3. 슬롯머신 AI 개발**
- 가장 많은 보상을 획득하는 AI 에이전트 만들기
- 카지노에 있는 5개의 슬롯머신 중 가장 높은 당첨 확률을 가진 머신을 스스로 찾아내는 AI 프로그램을 개발
- 각 슬롯머신은 서로 다른 당첨 확률을 가지고 있지만, AI 에이전트는 어떤 머신이 가장 유리한지 알지 못함
- 에이전트는 슬롯머신을 직접 선택하고, 당첨 여부를 확인하면서 각 머신의 당첨 확률을 추정해야 함
- 최종 목표는 1000번의 행동 동안 가능한 한 많은 보상을 획득하는 것

1. 슬롯머신 환경 구현
- Python과 NumPy를 사용
- 슬롯머신 총 5개
- 실제 당첨 확률은 [0.2, 0.5, 0.1, 0.8, 0.3]
- 슬롯머신을 선택하면 당첨 시 1, 실패 시 0의 보상을 반환하도록 구현
- AI 에이전트는 실제 당첨 확률을 직접 참조하여 행동을 선택해서는 안됨

````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
TRUE_PROBS = np.array([0.2, 0.5, 0.1, 0.8, 0.3])
N_STEPS = 1000

def bandit_experiment(strategy='epsilon', epsilon=0.1, seed=2026):
    rng = np.random.default_rng(seed)
    counts = np.zeros(5, dtype=int)
    estimates = np.zeros(5, dtype=float)
    rewards = []
    choices = []

    for t in range(N_STEPS):
        if t<5:
            action = t
        elif strategy == 'random':
            action = int(rng.integers(5))
        elif strategy == 'greedy':
            action = int(np.argmax(estimates))
        elif strategy == 'epsilon':
            action = int(rng.integers(5)) if rng.random() < epsilon else int(np.argmax(estimates))
        else:
            raise ValueError('unknown strategy')

        reward = float(rng.random() < TRUE_PROBS[action])
        counts[action] += 1
        estimates[action] += (reward - estimates[action]) / counts[action]
        rewards.append(reward)
        choices.append(action)
    return np.cumsum(rewards), counts, estimates
````

````python
results = {
    "Random" : bandit_experiment('random'),
    "Greedy" : bandit_experiment('greedy'),
    "ε=0" : bandit_experiment('epsilon', 0),
    "ε=0.1" : bandit_experiment('epsilon', 0.1),
    "ε=0.01" : bandit_experiment('epsilon', 0.01)
}
````

````python
for name, (cumulative, counts, estimates) in results.items():
    print(f"{name:8s} | 누적 보상={cumulative[-1]:.0f} | 선택 횟수={counts} | 추정={estimates.round(2)}")
plt.figure(figsize=(9, 4))
for name, (cumulative, _, _) in results.items():
    plt.plot(cumulative, label=name)
plt.xlabel("Number of actions")
plt.ylabel("Cumulative reward")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
````

밑에는 사전에 내가 코드를 짜본 것

````python
class SlotMachineEnv:
    def __init__(self):
        self.prob = np.array([0.2, 0.5, 0.1, 0.8, 0.3])

    def __len__(self):
        return len(self.prob)

    def pull(self, num):
        res = 1 if np.random.rand() < self.prob[num] else 0
        return res
````

````python
slotMachine = SlotMachineEnv()
total = 0
n = 10000

for i in range(n):
    total += slotMachine.pull(3)

print(total / n)
````

2. AI 에이전트 구현
- Random 전략: 5개의 슬롯머신 중 하나를 무작위로 선택
- Greedy 전략: 지금까지 획득한 보상을 기반으로 가장 높은 평균 보상을 가진 슬롯머신을 선택
- ε-greedy 전략: 일정 확률로 무작위 슬롯머신을 선택하고, 나머지 확률로 현재 가장 높은 평균 보상을 가진 슬롯머신을 선택(ε값은 0.1과 0.01을 각각 적용)

````python
class AiAgent:
    def __init__(self, size, eps):
        self.size = size # 슬롯 머신 개수
        self.counts = np.zeros(self.size)
        self.values = np.zeros(self.size)
        self.eps = eps

    def selectMachine(self):
        return np.random.randint(0,self.size)

    def update(self, num, res):
        self.counts[num] += 1
        self.values[num] = self.values[num] + (res- self.values[num]) / self.counts[num] # 평균 업데이트

    def greedy(self):
        return np.argmax(self.values)

    def eps_greedy(self):
        num = 1 if np.random.rand() < self.eps else 0

        if num == 1:
            return self.selectMachine()

        else:
            return self.greedy()
        
````

````python
agent = AiAgent(len(slotMachine), 0.1)
````

````python
for i in range(10000):
    num = agent.greedy()
    res = slotMachine.pull(num)
    agent.update(num,res)

print(agent.values)
````

````python
for i in range(10000):
    num = agent.eps_greedy()
    res = slotMachine.pull(num)
    agent.update(num,res)

print(agent.values)
````

# **4. 마르코프 성질**
"현재 상태와 현재 행동을 알면 다음 상태의 확률을 설명하는 데 과거 기록이 추가로 필요하지 않다."
- 어떤 확률 과정에서 미래 상태가 과거 전체의 이력에 의존하지 않고 오직 현재 상태에만 의존함
- 현재 상태가 과거 정보를 모두 요약하고 있다는 의미

### MP(Markov Process)
- 행동을 선택하는 주체가 없다는 것(예: 날씨)
```
오늘 날씨가 맑으면 내일은 아래와 같이 변한다고 가정
- 맑음 > 맑음(70%)
- 맑음 > 비(30%)
여기에는 누군가가 선택하는 행동이 없음
```
> 현재 상태에 따라 다음 상대가 확률적으로 결정

```
MP = (S, P)
- S: 상태(맑음, 비)
- P: 상태 전이확률
```

### MRP(Markov Reward Process)
- MP에 보상이라는 개념을 추가한 것
```
맑음 > 맑음 > 비 > 맑음
+10    +10   -5  +10
```

```
MRP = (S, P, R, γ)
- S: 상태
- P: 전이확률
- R: 보상
- γ: 할인율. 미래에 받을 보상을 현재 기준으로 얼마나 중요하게 평가할 것인지(γ값은 유지되는데, 미래로 갈수록 곱해져서 작은 값이 곱해짐)
```

### MDP(Markov Decision Process)
- MRP와의 큰 차이는 에이전트가 직접 행동을 선택할 수 있다는 것
- MRP에 행동 선택이 추가되어 정책을 비교할 수 있음
```
행동                성공확률        실패확률        성공보상        실패보상
위로 이동              80%           20%            +10            -10
아래로 이동            60%           40%             +5             -5

기대확률
0.8*(10) + 0.2*(-10) = 6
0.6*(5) + 0.4*(-5) = 1
```
> 위와 같은 환경에서 어떤 행동을 선택해야 미래에 많은 보상을 얻을 수 있는지 학습하는 것이 강화학습의 핵심

### 전이 확률

```
로봇이 "오른쪽으로 이동"을 명령했을 때
- 80%: 오른쪽으로 이동
- 10%: 미끄러져 위쪽으로 이동
- 10%: 제자리 유지
```
> 합계가 100%여야 함. 행동을 선택하더라도 다음 상태가 항상 확정되는 것은 아님

````python
transition = {"right": 0.8, "up": 0.1, "stay": 0.1}
assert np.isclose(sum(transition.values()), 1.0)
rng = np.random.default_rng(7)
samples = rng.choice(list(transition), size=10000, p=list(transition.values()))
for next_state in transition:
    print(next_state, "실험 비율:", round(np.mean(samples == next_state), 3))
````

### 정책, 상태 가치, 행동 가치
- 정책 π: 현재 상태에서 어떤 행동을 할지 결정하는 규칙. 확률적으로 행동을 고를 수도 있음
- 상태 가치 Vπ(s): 지금 이 상태에서 정책을 계속 따르면 앞으로 점수를 평균 얼마나 얻을까?
- 행동 가치 Qπ(s, a): 지금 이 상태에서 먼저 행동을 하고, 이후 정책을 따르면 앞으로 점수를 평균 얼마나 얻을까?

```
예: 교차로에서 "오른쪽"을 선택한 뒤 목적지에 도착할 가능성이 높다면 그 행동의 Q값이 높을 수 있음
V와 Q는 정책에 따라 달라질 수 있음
```

### 할인율 γ와 Return

```
게임에서 지금+10점, 다음 턴 +10점, 그 다음 턴 +10점을 받음
- γ=1: 10 + 10 + 10 = 30점(모두 같은 비중)
- γ=0.9: 10 + 9 + 8.1 = 27.1점
- γ=0.5: 10 + 5 + 2.5 = 17.5점
```
> γ는 "미래의 보상을 현재 평가에서 얼마나 중요하게 볼지"을 정함. γ가 1에 가까우면 미래를 더 오래 고려한다는 뜻이지, 미래가 현재보다 더 중요하다는 뜻은 아님

````python
rewards = [10, 10, 10]
for gamma in [0.0, 0.5, 0.9, 1.0]:
    discounted = sum((gamma ** t) * reward for t, reward in enumerate(rewards))
    print(f"gamma={gamma:.1f} -> return={discounted:.2f}")
````

# **5. 상태 전이와 할인 보상 계산**
- 일상생활에서 행동을 선택하는 AI 에이전트를 개발하려고 함
- 에이전트는 working(공부), gym(운동), home(귀가) 세 가지 상태를 가지며, 선택한 행동에 따라 상태가 변경되고 보상을 받음

```
현재 상태	    행동    	다음 상태   	보상
working	     working      working	      +3
working	       gym	        gym	          +5
working	      home	        home	     +12
gym	         working	  working	      +4
gym	          home	       home	         +12
```

### 구현 요구사항
- 상태 전이와 보상을 저장하는 딕셔너리를 작성
- 행동 목록과 할인율을 전달받는 함수를 구현
- 시작 상태는 working으로 설정
- 행동을 순서대로 실행하며 상태를 변경하고 보상을 저장
- 불가능한 행동을 선택하거나 종료 상태에서 추가행동을 실행하면 오류가 발생하도록 구현
- 아래 공식을 사용하여 할인 리턴을 계산
```
G = R1 + γR2 + γ²R3 + ...
```
- 할인율을 기본적으로 0.9를 사용
- 두 가지 경로를 실행하고 결과를 비교
    - working > working > home
    - gym > working > home

````python
TRANSITIONS = {
    ("working", "working"): ("working", 3),
    ("working", "gym"): ("gym", 5),
    ("working", "home"): ("home", 12),
    ("gym", "working"): ("working", 4),
    ("gym", "home"): ("home", 12)
}
````

````python
def evaluate_actions(actions, gamma=0.9):
    state, rewards = "working", []

    for action in actions:

        if state == "home":
            raise ValueError("종료 상태 이후 행동이 있습니다.")
        
        if (state, action) not in TRANSITIONS:
            raise ValueError(f"불가능한 행동: {state} > {action}")
        
        state, reward = TRANSITIONS[(state, action)]
        rewards.append(reward)

    if state != "home":
        print("주의: 아직 종료되지 않은 경로입니다.")
        
    return rewards, sum(gamma ** t * r for t, r in enumerate(rewards))
````

````python
for name,actions in {
    "A" : ["working", "working", "home"],
    "B" : ['gym', 'working', 'home']
}.items():
    rewards, G = evaluate_actions(actions)
    print(f'정책 경로 {name}: 보상={rewards}, 할인 리턴={G:.2f}')
````

````python
state_dict = {
    'working' : {'working' : 3, 'gym':5, 'home':12 },
    'gym' : {'working' : 4, 'home': 12 }
}

START_STATE = 'working'

def cal_reward(actions, sale=0.9):
    rewards = []
    result = 0

    cur_state = START_STATE

    for action in actions:
        rewards.append(state_dict[cur_state][action])
        cur_state = action

    
    for idx, num in  enumerate(rewards):
        result += sale ** idx * num

    return result
````

````python
aciton_1 = [ 'working' , 'working' , 'home' ]
action_2 = ['gym',  'working', 'home']

print('working > working > home: ',cal_reward(aciton_1))
print('gym > working > home: ',cal_reward(action_2))
````

# **6. 벨만 기대 방정식**
현재 상태의 가치를 계산할 때, 지금 받을 보상과 다음 상태에서 앞으로 받을 보상의 예상 가치를 합치는 공식

1. 학생이 게임을 하고 있다고 가정
- 보상: 특정 행동을 했을 때 즉시 얻는 점수
- 리턴: 지금부터 앞으로 얻을 보상들을 할인해서 합한 값
- 가치: 현재 상태에서 앞으로 얻을 리턴의 기댓값

```
           +2        +10
    시작    >   중간    >   종료

    단순 보상 합계: 2+10 =12
    할인율(0.9) 적용: 2+0.9*10 = 11
```
- 벨만 방정식이 필요한 이유
    - 위와 같은 예제는 보상이 두 번밖에 없어서 직접 더할 수 있음
    - 만약 게임이 100번, 1000번 진행된다면 매번 미래의 모든 보상을 직접 나열하는 것은 어려움
        - 벨만 방정식으로 해결
        - 현재 가치 = 지금 받을 보상 + 할인된 다음 상태의 가치
        - $V(s) = R + γV(s')$

```
    V(A) = 10
    V(S) = 2 + 0.9 * V(A)
    = 2 + 0.9*10 = 11 
```
> 미래 보상을 전부 다시 계산하지 않고, 다음 상태의 가치를 이용해 현재 가치를 계산하는 것

- 다음 상태가 확률적으로 달라질 경우

```
    게임에서 오른쪽 버튼을 눌렀는데, 항상 같은 칸으로 이동하지 않는다고 가정
    - 50% 확률로 좋은 칸에 이동
    - 50% 확률로 나쁜 칸에 이동
    - 좋은 칸의 미래 가치는 10점, 나쁜 칸의 미래 가치는 -2점

    기댓값을 계산
    10 * 0.5 + (-2) * 0.5 = 4(예상 가치)
    현재 보상 0이고, 할인율이 0.9라면
        0 + 0.9 * 4 = 3.6
    - 여기서 3.6은 실제로 받는 점수가 아님
    - 좋은 칸에 도착하면 할인된 미래 가치는 9점이고, 나쁜 칸에 도착하면 -1.8점 > 두 결과를 확률에 따라 평균내면 3.6이 됨
```

전체 벨만 방정식

$
\boxed{V^\pi(s)=\sum_a\pi(a|s)\sum_{s'}P(s'|s,a)[R(s,a,s')+\gamma V^\pi(s')]}
$

아래 3개를 합친 것

$R+\gamma V(s')$

$\sum_{s'}P(s'|s,a)[R+\gamma V(s')$

$\sum_a\pi(a|s)(\text{행동별 기대 가치})$

````python
good_value, bad_value = 10.0, -2.0
gamma = 0.9
expected_next = 0.5 * good_value + 0.5 * bad_value
print("다음 상태 가치의 평균:", expected_next)
print("지금 상태에서 기대하는 가치:", 0 + gamma * expected_next)
````

# **7. 집으로 돌아가는 길**
한 학생이 집으로 돌아가려고 합니다. 현재 위치는 A,B,C 중 하나이며, 이동하다 보면 집(Home)에 도착하거나 술집(Bar)에 도착할 수 있습니다. 각 위치에서 다음 위치로 이동할 확률은 미리 정해져 있으며, 학생은 이동 방향을 직접 선택할 수 없습니다. 이 환경에서 각 위치가 얼마나 유리한 상태인지 계산하는 프로그램을 작성하세요.

```
현재위치    다음 위치   이동 확률
A           Home        70%
A           B           30%
B           A           50%
B           C           50%
C           B           50%
C           Bar         50%
```
Home과 Bar에 도착하면 이동이 종료됩니다.

- Home에 도착하면 +10점
- Bar에 도착하면 -10점
- 그 외 이동은 0점
- 할인율은 0.9

- 상태 목록 STATES, 이동 규칙 NEXT, 종료 상태 TERMINAL을 정의
- 다음 위치가 Home이면 +10, Bar이면 -10, 그 외에는 0을 반환하는 함수를 작성
- 모든 상태의 가치를 0으로 초기화한 후 벨만 기대 방정식을 이용해서 각 상태의 가치를 계산
- 최대 100회 반복. 매 반복마다 A, B, C의 가치를 갱신. 이전 반복과 현재 반복의 최대 가치 차이를 계산. 반복에 따른 가치 변화를 출력

````python
STATES = ['Home','A','B','C','Bar']
NEXT = {
    'A' : [('Home', 0.7), ('B', 0.3)],
    'B' : [('A',0.5),('C',0.5)],
    'C': [('B',0.5), ('Bar',0.5)]
}
TERMINAL = {'Home', 'Bar'}
````

````python
def arrival_reward(next_state):
    return 1.0 if next_state == 'Home' else -1.0 if next_state == "Bar" else 0.0
````

````python
# 상태 가치 함수를 저장
V = {state: 0.0 for state in STATES}
for iteration in range(1, 101):
    new_V = V.copy()
    for state in ['A', 'B', 'C']:
        # state: 현재 상태
        # nxt: 다음 상태
        # prob: 전이 확률
        # arrival_reward(nxt): 이동 시 보상
        # V[nxt]: 다음 상태의 가치
        new_V[state] = sum(prob * (arrival_reward(nxt)+V[nxt]) for nxt, prob in NEXT[state]) # 벨만 기대 방정식 적용

    # 이전 반복과 현재 반복의 상태 가치가 얼마나 달라졌는지 비교
    delta = max(abs(new_V[s] - V[s]) for s in STATES)

    # 이전 반복에서 계산한 새로운 상태 가치를 다음 반복에서 사용할 수 있도록 갱신
    V = new_V

    if iteration in [1, 2, 5, 10, 20, 50, 100]:
        print(f'{iteration:3d}회: A={V['A']:+.4f}, B={V['B']:+.4f}, C={V['C']:+.4f}')

    if delta < 1e-12: # 상태가치 변화량이 미미한 경우 즉시 종료
        break

# assert: 특정 조건이 참인지 확인
# allclose: 두 배열의 값이 허용 오차 범위 내에서 거의 같은지 확인
# atol: 절대 허용 오차를 설정
assert np.allclose([V['A'], V['B'], V['C']], [0.75, 1/6, -5/12], atol=1e-6) # assert 일부러 에러 내는 코드
````

### **8. 몬테카를로 방법**
몬테카를로(Monte Carlo) 방법은 무작위 표본을 반복적으로 생성하고, 그 결과의 평균을 이용하여 어떤 값의 기댓값을 추정하는 방법
```
위 예제를 A에서 5번 실행

실행    최종 도착    보상
1       Home        +1
2       Home        +1
3       Bar         -1
4       Home        +1
5       Home        +1

평균 보상: 0.6
```
> 5번만 실행했기 때문에 정확하지 않을 수 있지만, 실험 횟수를 10000번 정도로 늘리면 평균값은 이 문제의 이론적 가치인 0.5에 가까워지는 경향이 있음

몬테카를로
$V_\pi(s)\approx\frac{1}{N(s)}\sum_{i=1}^{N(s)}G_i(s)$

리턴 계산
$G_t=R_{t+1}+\gamma R_{t+2}+\gamma^2R_{t+3}+\cdots$

### 벨만 방정식과 몬테카를로의 차이
1. 벨만 방정식
- 계산 원리: 전이 확률로 기대값 계산
- 전이 확률: 필요
- 에피소드: 실행할 필요 없음
- 결과: 수렴한 상태 가치
- 특징:  모델 기반

2. 몬테카를로
- 계산 원리: 실제 경험의 리턴 평균
- 전이 확률: 필요 없음
- 에피소드: 종료까지 실행
- 결과: 경험 기반 추정 가치
- 특징: 모델 프리


````python
def random_walk_episode(start, rng):
    state = start
    
    while state not in TERMINAL:
        options, probs = zip(*NEXT[state])
        state = str(rng.choice(options, p=probs))

    return arrival_reward(state)
````

````python
rng = np.random.default_rng(2026)

for start in ['A', 'B', 'C']:
    scores = [random_walk_episode(start, rng) for _ in range(10000)]
    print(f'{start}: 몬테카를로 평균={np.mean(scores):+.4f} | 벨만 계산={V[start]:+.4f}')
````

````python
SIZE = 5
HOME = (0,0)
BAR = (0,4)
GAMMA = 0.95
ACTIONS = {
    'UP': (-1,0),
    'LEFT':(0,-1),
    'RIGHT':(0,1),
    'DOWN':(1,0)
}
ACTION_NAMES = ['UP','DOWN','LEFT','RIGHT']
ARROWS = {'UP': "↑", "DOWN":"↓","LEFT":"←","RIGHT":"→",None:"·"}
````

````python
def grid_step(state,action):
    if state in (HOME,BAR):
        return state, 0.0, True
    dr, dc = ACTIONS[action]
    row, col = state
    next_state = (min(max(row+dr, 0), SIZE-1), min(max(col+dc, 0), SIZE-1))
    if next_state == HOME:
        reward = 1.0
    elif next_state == BAR:
        reward = -1.0
    else:
        reward = 0.0
    done = next_state in (HOME, BAR)
    return next_state, reward, done
````

````python
def evaluate_random_policy(gamma=GAMMA, tol=1e-10, max_iter=10000):
    values = np.zeros((SIZE,SIZE))

    for iteration in range(max_iter):
        updated = values.copy()

        for row in range(SIZE):
            for col in range(SIZE):
                state = (row,col)
                if state in (HOME,BAR):
                    continue
                action_values = []

                for action in ACTION_NAMES:
                    nxt, reward, done = grid_step(state, action)
                    future = 0.0 if done else values[nxt]
                    action_values.append(reward+gamma*future)
                updated[state] = np.mean(action_values)
        delta = np.max(np.abs(updated-values))
        values = updated
        if delta < tol:
            return values, iteration + 1
    raise RuntimeError('랜덤 정책 평가가 수렴하지 않음')
````

# **10. 정책 개선**
- 랜덤 정책의 가치는 '마구 걸으면 이 칸이 얼마나 좋은가?'임
- 그 숫자를 이용하면 각 칸에서 더 나은 화살표를 고를 수 있음
- Q(s,a) = r + γ * (종료면 0, 아니면 V(s'))
- 개선된 정책은 각 칸에서 이 Q가 가장 큰 행동을 고름. 점수가 같으면 up, down, left, right 중 먼저 나온 행동을 고름

````python
def action_scores(state, values, gamma=GAMMA):
    scores = []

    for action in ACTION_NAMES:
        nxt, reward, done = grid_step(state, action)
        future = 0.0 if done else values[nxt]
        scores.append(reward + gamma * future)

    return scores
````

````python
def greedy_policy_from(values, gamma = GAMMA):
    policy = np.empty((SIZE, SIZE), dtype=object)

    for row in range(SIZE):
        for col in range(SIZE):
            state = (row, col)
            if state in (HOME, BAR):
                policy[state] = None
            else:
                policy[state] = ACTION_NAMES[int(np.argmax(action_scores(state,values,gamma)))]
    return policy
````

````python
one_step_policy = greedy_policy_from(random_values)
print('랜덤 정책 가치로 한 번만 개선한 화살표')
show_policy(one_step_policy)
````

# **11. 정책 반복**
정책 반복은 두 단계를 정책이 변하지 않을 때까지 반복
1. 정책 평가: 지금 화살표를 고정하고, 그 화살표만 따르는 V가 수렴할 때까지 갱신
2. 정책 개선: 그 V로 각 칸의 화살표를 다시 그림

> 평가 > 개선 > 비교 > 다시 평가(반복)

````python
def evaluate_policy(policy, gamma=GAMMA, tol=1e-10, max_iter=10000):
    values = np.zeros((SIZE,SIZE))
    for iteration in range(max_iter):
        updated = values.copy()
        for row in range(SIZE):
            for col in range(SIZE):
                state = (row,col)
                if state in (HOME, BAR):
                    continue
                nxt, reward, done = grid_step(state, policy[state])
                future = 0.0 if done else values[nxt]
                updated[state] = reward + gamma * future
        delta = np.max(np.abs(updated-values))
        values = updated
        if delta < tol:
            return values, iteration + 1
    raise RuntimeError('정책 평가가 수렴되지 않았음')
````

````python
def policies_equal(left,right):
    return all(left[r,c]==right[r,c] for r in range(SIZE) for c in range(SIZE))
````

````python
def policy_iteration(gamma=GAMMA, max_outer=20):
    policy = np.full((SIZE,SIZE), 'RIGHT', dtype=object)
    policy[BAR] = None
    for outer in range(1, max_outer+1):
        values, eval_iters = evaluate_policy(policy, gamma=gamma)
        improved = greedy_policy_from(values,gamma=gamma)
        print(f"바깥 {outer}회 | 평가 반복={eval_iters}")
        show_policy(improved)
        if policies_equal(policy,improved):
            return values,improved,outer
        policy = improved
    raise RuntimeError('정책 반복이 끝나지 않음')
````

````python
pi_values, pi_policy, pi_rounds = policy_iteration()
print('정책 반복 최종 가치')
show_policy(pi_values)
````
